```python
string_columns = df.select_dtypes(include=['object', 'string']).columns.tolist()
print("String columns:", string_columns)
print("Number of string columns:", len(string_columns))
```

```python
X = df.drop(columns=['CLASS', 'ID'])
y = df['CLASS']
```

```python
# Check for infinite values
has_inf = X.map(lambda x: np.isinf(x) if np.issubdtype(type(x), np.number) else False).any()
inf_columns = has_inf[has_inf].index.tolist()
print("Columns with infinite values:", inf_columns)

# Check for very large values
large_values = X.select_dtypes(include=[np.number]).abs() > np.finfo(np.float64).max
large_columns = large_values.any()[large_values.any()].index.tolist()
print("Columns with values too large for float64:", large_columns)
```

```python
# replacing the infinite values with NaN
X = X.replace([np.inf, -np.inf], np.nan)

# Check for infinite values
has_inf = X.map(lambda x: np.isinf(x) if np.issubdtype(type(x), np.number) else False).any()
inf_columns = has_inf[has_inf].index.tolist()
print("Columns with infinite values:", inf_columns)

# Check for very large values
large_values = X.select_dtypes(include=[np.number]).abs() > np.finfo(np.float64).max
large_columns = large_values.any()[large_values.any()].index.tolist()
print("Columns with values too large for float64:", large_columns)
```
